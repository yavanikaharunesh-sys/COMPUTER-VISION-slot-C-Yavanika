```python

from google.colab import files
from IPython.display import Video, display
import cv2, os, subprocess

# Upload input video
uploaded = files.upload()
name = next(iter(uploaded))

cap = cv2.VideoCapture(name)
if not cap.isOpened():
    raise ValueError("Could not open video. Please upload a valid MP4 file.")

fps = cap.get(cv2.CAP_PROP_FPS)
if fps <= 0:
    fps = 24

w = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
h = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

# Temporary AVI files
slow = cv2.VideoWriter("slow_temp.avi",
    cv2.VideoWriter_fourcc(*'MJPG'), max(fps / 2, 1), (w, h))
fast = cv2.VideoWriter("fast_temp.avi",
    cv2.VideoWriter_fourcc(*'MJPG'), fps, (w, h))

i = 0
while True:
    ret, frame = cap.read()
    if not ret:
        break
    slow.write(frame)
    if i % 2 == 0:
        fast.write(frame)
    i += 1

cap.release()
slow.release()
fast.release()

if i == 0:
    raise ValueError("No frames were read from the video.")

print("Frames processed:", i)

# Convert to browser-friendly MP4
for src, dst in [("slow_temp.avi", "slow.mp4"),
                 ("fast_temp.avi", "fast.mp4")]:
    subprocess.run([
        "ffmpeg", "-y", "-i", src,
        "-vcodec", "libx264", "-pix_fmt", "yuv420p",
        "-movflags", "+faststart", dst
    ], check=True, stdout=subprocess.DEVNULL,
       stderr=subprocess.DEVNULL)

print("SLOW MOTION OUTPUT")
display(Video("slow.mp4", embed=True, width=500))

print("FAST MOTION OUTPUT")
display(Video("fast.mp4", embed=True, width=500))
```
