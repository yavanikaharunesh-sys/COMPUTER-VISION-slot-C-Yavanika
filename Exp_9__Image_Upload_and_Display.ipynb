```python

from google.colab import files
import cv2
import matplotlib.pyplot as plt

f = files.upload()
img = cv2.imread(next(iter(f)))
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title('Uploaded Image')
plt.axis('off')
plt.show()
```
