```python

from google.colab import files
import cv2
import matplotlib.pyplot as plt

f = files.upload()
img = cv2.imread(next(iter(f)), 0)
x = cv2.convertScaleAbs(cv2.Sobel(img, cv2.CV_64F, 1, 0, ksize=3))
y = cv2.convertScaleAbs(cv2.Sobel(img, cv2.CV_64F, 0, 1, ksize=3))
out = cv2.addWeighted(x, 0.5, y, 0.5, 0)
plt.subplot(1,3,1); plt.imshow(x, cmap='gray'); plt.title('Sobel X'); plt.axis('off')
plt.subplot(1,3,2); plt.imshow(y, cmap='gray'); plt.title('Sobel Y'); plt.axis('off')
plt.subplot(1,3,3); plt.imshow(out, cmap='gray'); plt.title('Combined'); plt.axis('off')
plt.show()
```
