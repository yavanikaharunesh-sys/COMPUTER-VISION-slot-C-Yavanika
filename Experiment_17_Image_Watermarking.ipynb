```python

from google.colab import files
import cv2
import matplotlib.pyplot as plt

f = files.upload()
img = cv2.imread(next(iter(f)))
out = img.copy()
cv2.putText(out, 'SIMATS', (20,60), cv2.FONT_HERSHEY_SIMPLEX, 1, (255,255,255), 2)
plt.subplot(1,2,1); plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB)); plt.title('Original'); plt.axis('off')
plt.subplot(1,2,2); plt.imshow(cv2.cvtColor(out, cv2.COLOR_BGR2RGB)); plt.title('Watermarked'); plt.axis('off')
plt.show()
```
