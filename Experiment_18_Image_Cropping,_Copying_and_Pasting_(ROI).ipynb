```python

from google.colab import files
import cv2
import matplotlib.pyplot as plt

f = files.upload()
img = cv2.imread(next(iter(f)))
h, w = img.shape[:2]
roi = img[h//4:h//2, w//4:w//2].copy()
out = img.copy()
rh, rw = roi.shape[:2]
out[h-rh:h, w-rw:w] = roi
plt.subplot(1,3,1); plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB)); plt.title('Original'); plt.axis('off')
plt.subplot(1,3,2); plt.imshow(cv2.cvtColor(roi, cv2.COLOR_BGR2RGB)); plt.title('Cropped ROI'); plt.axis('off')
plt.subplot(1,3,3); plt.imshow(cv2.cvtColor(out, cv2.COLOR_BGR2RGB)); plt.title('Pasted ROI'); plt.axis('off')
plt.show()
```
