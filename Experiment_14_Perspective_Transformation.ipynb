```python

from google.colab import files
import cv2, numpy as np
import matplotlib.pyplot as plt

f = files.upload()
img = cv2.imread(next(iter(f)))
h, w = img.shape[:2]
src = np.float32([[0,0],[w-1,0],[0,h-1],[w-1,h-1]])
dst = np.float32([[w*.1,h*.1],[w*.9,0],[0,h*.9],[w*.9,h*.9]])
m = cv2.getPerspectiveTransform(src, dst)
out = cv2.warpPerspective(img, m, (w,h))
plt.subplot(1,2,1); plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB)); plt.title('Original'); plt.axis('off')
plt.subplot(1,2,2); plt.imshow(cv2.cvtColor(out, cv2.COLOR_BGR2RGB)); plt.title('Perspective'); plt.axis('off')
plt.show()
```
